```python
# welcome to banking program 

def show_balance(balance) :
    print(f"your balance is {balance :.2f}")
    
def deposit() :
    amount = float(input("enter your money to deposit"))
    if amount < 0 :
        print("that is negative number")
        return 0
    else :
        return amount
    
def withdraw(balance) :
    amount = float(input("enter your number to withdraw"))
    if amount > balance :
        print("that is not allow")
        return  0 
    elif  amount < 0 :
        print("that is negative number ") 
        return 0  
    else :
        return amount            

def main() :
    balance = 0
    is_running = True
    while is_running :
        print("##### welcome to banking program #####")
        print ("1. show balance")
        print("2. deposit")
        print("3. wihtdraw")
        print("4. exit of program")
        choice = int(input("enter your process (1-4)"))
        match choice :
            case 1 :
                show_balance(balance)
            case 2 :
                balance += deposit()
            case 3 :
                balance -= withdraw(balance)
            case 4 :
                is_running = False   
            case _:
                print("that is choise not a valid")

    print(" & thank you have a nice day & ")
if __name__ == "__main__" :
    main()                
                          
```

```python
import random
def spin_row() :
    your_emoji =["❤" , "🌹" , "🔴"]
    result=[]
    
    for emoji in range(3) :
        result.append(random.choice(your_emoji))
    return result    
        
def print_row(row) :
    print(" | ".join(row))

def get_payout(row , bet) :
    if row[0] == row[1] == row[2] :
       if row[0] == "❤" :
           return bet + 10
       elif row[0] == "🌹" :
           return bet + 20
       elif row[0] == "🔴" :
           return bet * 4
    return 0     

def main() :
    balance = 100
    print(" ##### hello in python slot ##### ")
    print(" ##### your emoji : ❤ 🌹 🔴 ####")
    
    while balance > 0 :
        print(f"your balance is : {balance : .2f}")
        bet = input("enter your money to play : ")
        if not bet.isdigit() :
            print("please inter your number")
            continue
        bet =int(bet)
        if bet > balance :
            print("that is not allow !!")
            continue
        if bet <= 0 :
            print("that is negative !!")
            continue
        balance = balance - bet  
        row = spin_row()
        print("spining .....")    
        print_row(row)  
        payout = get_payout(row , bet)
        
        if payout > 0 :
            print(f"you are win ${payout}")
        else :
            print("good luck")   
        balance = balance + payout    
    
if __name__ =="__main__" :
    main()    
```

```python

import random
words = ("apple" , "python" , "kareem" , "bassant" , "mohamed" , "labtob")

hangman_art = {0 : ("   " ,
                    "   " ,
                    "   ") ,
               1 : (" o " ,
                    "   " ,
                    "   ") ,
               2 : (" o " ,
                    " | " ,
                    "   ") ,
               3 : (" o " ,
                    " | " ,
                    "/  ") ,
               4 : (" o " ,
                    " | " ,
                    "/ \\ ") ,
               5 : (" o " ,
                    "/| " ,
                    "/ \\ ") ,
               6 : (" o " ,
                    "/|\\ " ,
                    "/|\\ ") ,}

def display_man (wrong_guesses) :
    for line in hangman_art [wrong_guesses] :
        print(line)

def display_hint (hint) :
    print(" ".join(hint))

def display_answer (answer) :
    print(" ".join(answer))

def main () :
    answer = random.choice(words)
    hint = ["_"] * len(answer)
    wrong_guesses = 0
    guessed_letters = set()
    is_running = True
    while is_running :
        display_man(wrong_guesses)
        display_hint(hint)
        guess = input("enter your letter : ").lower()
        
        if len(guess) != 1 or not guess . isalpha() :
             print("you should write one letter !")
             continue
        
        if guess in guessed_letters :
             print(f"the {guess} is already guessed !!")
             continue
        
        guessed_letters.add(guess)
        if guess in answer :
            for i in range(len(answer)) :
                if answer [i] == guess :
                     hint [i] = guess
        else :
                     wrong_guesses += 1    
        
        if "_" not in hint :
             display_man(wrong_guesses)
             display_answer(answer)
             print("congratulation $ you are win $ ") 
             is_running = False                   
                    
    
if __name__ == "__main__":
    main()
    
```
